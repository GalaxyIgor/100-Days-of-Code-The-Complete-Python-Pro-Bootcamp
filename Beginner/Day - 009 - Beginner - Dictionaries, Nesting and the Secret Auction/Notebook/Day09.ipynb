## 68. The Python Dictionary: Deep Dive
dictionaries = {key: value}

Note that "weight" does not exist in the first dictionary, but after we add it, "weight" exists.

If you want to change the value of a key, follow the "weight" example.

You can loop through a dictionary; follow the example below.

Note that, in the end, example = {} turns the dictionary into an empty dictionary.
### Example:

```python
example = {
    "name": "Angela",
    "age": 30,
    "height": 1.7
}

print("\nThis is the example dictionary")
print(example)

print("\nThis is the name at dictionary")
print(example["name"])

print("\nThis is the age at dictionary")
print(example["age"])

print("\nThis is the height at dictionary")
print(example["height"])

example["weight"] = 70

print("\nThis is the weight at dictionary")
print(example["weight"])

print("\nThis is the example dictionary with a new key")
print(example)

example["weight"] = 100

print("\nThis is the example dictionary with a new value for a existing key")
print(example)

print("\nThis is a for loop")
for thing in example:
    print(thing)

print("\nThis is a for loop with key:value")
for key in example:
    print(key)
    print(example[key])

example = {}

print("\nThis is the example of empty dictionary")
print(example)
```

## Coding Exercise 9: Grading Program
You have access to a database of student_scores in the format of a dictionary. The keys in student_scores are the names of the students and the values are their exam scores.

Write a program that converts their scores to grades.

By the end of your program, you should have a new dictionary called student_grades that should contain student names as keys and their assessed grades for values.

The final version of the student_grades dictionary will be checked.

**DO NOT** modify lines 1-7 to change the existing student_scores dictionary.


This is the scoring criteria:
- Scores 91 - 100: Grade = "Outstanding"
- Scores 81 - 90: Grade = "Exceeds Expectations"
- Scores 71 - 80: Grade = "Acceptable"
- Scores 70 or lower: Grade = "Fail"

```python
student_scores = {
    'Harry': 88,
    'Ron': 78,
    'Hermione': 95,
    'Draco': 75,
    'Neville': 60
}

student_grades ={}

for student, score in student_scores.items():
    if score >= 91 and score <= 100:
        student_grades[student] = "Outstanding"
    elif score >= 81 and score <=90:
        student_grades[student] = "Exceeds Expectations"
    elif score >= 71 and score <= 80:
        student_grades[student] = "Acceptable"
    elif score <= 70:
        student_grades[student] = "Fail"
```

## 69. Nesting Lists and Dictionaries
Nesting = Nesting means placing one data structure inside another.

```python
capitals = {
    "France": "Paris",
    "Germany": "Berlin"
}

print("\nThis is the capital dictionary")
print(capitals)

# Nested list in dictionary
travel_log = {
    "France": ["Paris", "Lille", "Dijon"],
    "Germany": ["Stuttgart", "Berlin"]
}

print("\nFrance travel log:")
print(travel_log["France"])

print("\nGermany travel log:")
print(travel_log["Germany"])

print("\nGermany first city")
print(travel_log["Germany"][0])

print("\nGermany second city")
print(travel_log["Germany"][1])

print("\n This is a Nested List")
nested_list = ["A", "B", ["C", "D"]]
print(nested_list)
```
