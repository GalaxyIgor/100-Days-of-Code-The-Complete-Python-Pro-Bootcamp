## 73. Functions with Outputs
In Py, you can create functions with:

```python
def my_function():
    print("Hello")

my_function()
```

And you can do with inputs:

```python
def my_function(number1, number2):
    print(number1 + number2) # 1+2 = 3

my_function(1,2)
```

And finally you can do with outputs:

```python
def my_function():
    result = 1+2
    return result # 1+2 = 3

output = my_function()
print(output)
```

```python
def format_name(f_name, l_name):
    f_name = f_name.title()
    l_name = l_name.title()
    return f_name + " " + l_name

format_String = format_name("anGela","yu")
print(format_String)

output = len("Angela")
print(output)
```
